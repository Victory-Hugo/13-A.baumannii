```python
df_1_毒力因子_file = "/mnt/d/1-ABaumannii/1-注释汇总/1-NCBI-Origin/All_Samples_毒力因子_阈值.csv"
df_2_毒力因子_file = "/mnt/d/1-ABaumannii/1-注释汇总/2-NCBI-Sequence/All_Samples_毒力因子_阈值.csv"
df_1_抗生素耐药_file = "/mnt/d/1-ABaumannii/1-注释汇总/1-NCBI-Origin/All_Samples_抗生素耐药.csv"
df_2_抗生素耐药_file = "/mnt/d/1-ABaumannii/1-注释汇总/2-NCBI-Sequence/All_Samples_抗生素耐药.csv"
df_1_生物杀灭_file = "/mnt/d/1-ABaumannii/1-注释汇总/1-NCBI-Origin/All_Samples_生物杀灭抵抗_阈值.csv"
df_2_生物杀灭_file = "/mnt/d/1-ABaumannii/1-注释汇总/2-NCBI-Sequence/All_Samples_生物杀灭抵抗_阈值.csv"
```

```python
毒力因子_合并_file = "/mnt/d/1-ABaumannii/1-注释汇总/毒力因子_合并.csv"
抗生素耐药_合并_file = "/mnt/d/1-ABaumannii/1-注释汇总/抗生素耐药_合并.csv"
生物杀灭_合并_file = "/mnt/d/1-ABaumannii/1-注释汇总/生物杀灭抵抗_合并.csv"
```

```python
df_1_毒力因子 = pd.read_csv(df_1_毒力因子_file)
df_2_毒力因子 = pd.read_csv(df_2_毒力因子_file)
```

```python
import re
import io

def preprocess_csv_content(file_path):
    """
    预处理CSV文件，将多余字段中的逗号替换为空格
    """
    with open(file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    if not lines:
        return ""
    
    # 获取头行的字段数
    header = lines[0].strip()
    expected_fields = len(header.split(','))
    print(f"期望字段数: {expected_fields}")
    
    processed_lines = [header]
    problem_lines = 0
    
    for i, line in enumerate(lines[1:], 1):
        line = line.strip()
        fields = line.split(',')
        
        if len(fields) > expected_fields:
            # 有多余的字段，需要处理最后几个字段
            excess_count = len(fields) - expected_fields
            # 将多余的字段合并到倒数第二个字段中，并用空格替换逗号
            extra_content = ' '.join(fields[expected_fields-1:])
            # 重新构建行
            new_fields = fields[:expected_fields-1] + [extra_content]
            processed_line = ','.join(new_fields)
            processed_lines.append(processed_line)
            problem_lines += 1
            if problem_lines <= 5:  # 只打印前5个问题行的信息
                print(f"处理问题行 {i+1}: 字段数 {len(fields)} -> {expected_fields}")
        else:
            processed_lines.append(line)
    
    print(f"总共处理了 {problem_lines} 个问题行")
    return '\n'.join(processed_lines)

def read_csv_with_preprocessing(file_path):
    """
    使用预处理读取CSV文件
    """
    try:
        processed_content = preprocess_csv_content(file_path)
        df = pd.read_csv(io.StringIO(processed_content))
        print(f"成功读取 {len(df)} 行数据")
        return df
    except Exception as e:
        print(f"读取文件 {file_path} 时出错: {e}")
        return None

# 重新读取所有文件
print("正在读取抗生素耐药数据...")
df_1_抗生素耐药 = read_csv_with_preprocessing(df_1_抗生素耐药_file)

print("\n正在读取第二个抗生素耐药数据...")
df_2_抗生素耐药 = read_csv_with_preprocessing(df_2_抗生素耐药_file)

print("\n正在读取生物杀灭数据...")
df_1_生物杀灭 = read_csv_with_preprocessing(df_1_生物杀灭_file)

print("\n正在读取第二个生物杀灭数据...")
df_2_生物杀灭 = read_csv_with_preprocessing(df_2_生物杀灭_file)
```

```python
df_毒力因子 = pd.concat([df_1_毒力因子, df_2_毒力因子], ignore_index=True)
df_抗生素耐药 = pd.concat([df_1_抗生素耐药, df_2_抗生素耐药], ignore_index=True)
df_生物杀灭 = pd.concat([df_1_生物杀灭, df_2_生物杀灭], ignore_index=True)
```

```python
# 将第一列的"_vs_生物杀灭.txt"删去
df_生物杀灭.iloc[:, 0] = df_生物杀灭.iloc[:, 0].str.replace("_vs_生物杀灭.txt", "", regex=False)
```

```python
# 将第一列的"_AMRFinder.tsv"删去
df_抗生素耐药.iloc[:, 0] = df_抗生素耐药.iloc[:, 0].str.replace("_AMRFinder.tsv", "", regex=False)
```

```python
# 将第一列的"_vs_VFDB.txt"删去
df_毒力因子.iloc[:, 0] = df_毒力因子.iloc[:, 0].str.replace("_vs_VFDB.txt", "", regex=False)
```

```python
df_毒力因子.to_csv(毒力因子_合并_file, index=False)
df_抗生素耐药.to_csv(抗生素耐药_合并_file, index=False)
df_生物杀灭.to_csv(生物杀灭_合并_file, index=False)
```
